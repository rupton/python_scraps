```python
from datetime import datetime

now = datetime.now()
print(now)
print(now.date())
print(now.time())
```

```python
from time import time, ctime, localtime
time_t = time()
print(time_t)
time_c = ctime(time_t)
print(time_c)
time_l = localtime(time_t)
print(time_l)
```

```python
from time import time
print(time())
time_t = time()
time_ml = int(time_t * 1000)
print("Current time in milliseconds: ", time_ml)
```

```python
Print date in current timezone
```

```python
from datetime import datetime, timezone
date_t = datetime.now()
print(date_t)
print(date_t.now((timezone.utc)))
```
